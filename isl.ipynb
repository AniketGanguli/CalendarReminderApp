```python
import zipfile
import os

# Unzip archive.zip into SignData folder
with zipfile.ZipFile("archive.zip", 'r') as zip_ref:
    zip_ref.extractall("SignData")

# Verify folder structure
print("Top-level folders inside 'SignData':", os.listdir("SignData"))
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Define image size and paths
image_size = (64, 64)
batch_size = 32
train_path = "SignData/isl_data_grey_split/train"
test_path = "SignData/isl_data_grey_split/test"

# Data Generators
train_datagen = ImageDataGenerator(rescale=1./255)
test_datagen  = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    train_path,
    target_size=image_size,
    color_mode='grayscale',
    batch_size=batch_size,
    class_mode='categorical',
    shuffle=True
)

test_generator = test_datagen.flow_from_directory(
    test_path,
    target_size=image_size,
    color_mode='grayscale',
    batch_size=batch_size,
    class_mode='categorical',
    shuffle=False
)

# Build CNN Model
model = models.Sequential([
    layers.Input(shape=(64, 64, 1)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(train_generator.num_classes, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Train model
history = model.fit(
    train_generator,
    validation_data=test_generator,
    epochs=10
)
```

```python
model.save("sign_language_model.keras")

```
